```python
# 9.	Change all column names to lowercase, and replace spaces with underscores.
df.columns = df.columns.str.lower().str.replace(' ', '_')
df.columns
#  first columns are converted to lowercase and then space is replaced with underscore
```

```python
# 12.	Replace the text values NA, N/A, null and - with proper missing values.
import numpy as np
missing_values = ['NA', 'N/A', 'null', '-']

before_null_count = df.isnull().sum().sum()

df = df.replace(missing_values, np.nan)

after_null_counts = df.isnull().sum().sum()
print(after_null_counts)
print("Replaced values: ", after_null_counts - before_null_count)
# after replacing null values increase by 1782
```

```python
# 25.	Convert the hours logged column to a number.
df['hours_logged'] = pd.to_numeric(df['hours_logged'], errors='coerce')
df['hours_logged'].dtype

# here coerce parse the invalid values that cannot be converted to numberes ot NaN
```

```python
# 26.	Convert the billing rate column to a number. Remove the $ sign first.
df["billing_rate_(usd)"] = df['billing_rate_(usd)'].str.replace('$','')
df['billing_rate_(usd)'] = pd.to_numeric(df['billing_rate_(usd)'], errors='coerce')
df['billing_rate_(usd)']

# this first remove $ sign and convert column to number. 
```

```python
# 27.	Convert the CSAT score column to a number.
df['csat_score'] = pd.to_numeric(df['csat_score'], errors='coerce')
df['csat_score'].dtype

# csat_score converted to float64
```

```python
# 28.	Print the average, the smallest and the largest value of hours logged.
print("average: ", df['hours_logged'].mean())
print("smallest: ", df['hours_logged'].min())
print("largest: ", df['hours_logged'].max())

# average of hours_logged in 64.06 smallest is -12.5 and largest is 9888.01
```

```python
# 29.	Sort the table by hours logged from highest to lowest and show the top 10 rows.
df.sort_values(by='hours_logged', ascending = False).head(10)

# top 10 values by hours_logged from higest to loweest
# top hours logged is 9888.01
```

```python
# 30.	Show the 5 tickets with the highest billable amount.

df['billable_amount_usd'] = df['billable_amount_usd'].str.replace('$','')
df['billable_amount_usd']
df['billable_amount_usd'] = pd.to_numeric(df['billable_amount_usd'], errors='coerce')
df.sort_values(by = "billable_amount_usd", ascending = False).head(5)

# first replace $ and then convert to numaric and sort 
# top billable amount is 2564625.0
```

```python
# 31.	Remove the extra spaces from the client name column.
df['client_name'] = df['client_name'].str.strip()
df['client_name']

# str.strip() is used to remove any leading or trailing whitespaces 
```

```python
# 32.	Change the client name column to title case.
df['client_name'] = df['client_name'].str.title()
df['client_name'].head()

# str.title() is used to convert to title case
```

```python
# 33.	Convert the ticket created column to a date.
df['ticket_created_at'] = pd.to_datetime(df['ticket_created_at'], errors = 'coerce')
df['ticket_created_at'].dtype
df['ticket_created_at']

# pd.to_datetime() coverts to datetime format yyyy-mm-dd hh:mm:ss
```

```python
# 34.	Add a new column that holds the year of the ticket.
df['ticket_year'] = df['ticket_created_at'].dt.year
df[['ticket_created_at','ticket_year']].head()

#  ticket_year column is created to hold the year of the ticket.
```

```python
# 35.	Add a new column that holds the month of the ticket.
df['ticket_month'] = df['ticket_created_at'].dt.month
df[['ticket_created_at','ticket_month']].head()
df['ticket_month'].dtype

# ticket_month column is created to hold month. By default data type of new column is float 
```

```python
# 36.	Count how many tickets were created in each year.
df['ticket_year'].value_counts()

# 2023 has highest tickets
```

```python
# 37.	Add a new column called total equal to billing rate multiplied by hours logged.
df.columns
df['total_billing_rate'] = df['billing_rate_(usd)']*df['hours_logged']
df[['billing_rate_(usd)','hours_logged','total_billing_rate']].head()

# new column is created with total_billing_rate
```

```python
# 38.	Fill the missing values in the CSAT score column with the average CSAT score.
# df.columns
avg_csat_score = df['csat_score'].mean()
print("average csat score: ", avg_csat_score)
df['csat_score'] = df['csat_score'].fillna(avg_csat_score)
df['csat_score'].isnull().sum()
df['csat_score']
# null values are replaced with the mean value 
```

```python
# 39.	Drop the rows where the client name is missing.
rows = df.shape[0]
df.dropna(subset=['client_name'], inplace=True)
print("dropped rows: ",rows - df.shape[0])
print(df['client_name'].isnull().sum())
# there is no null values in client_name column
```

```python
# 40.	Group by service line and find the average hours logged.
service_line_by_hours = df.groupby('serviceline')['hours_logged'].mean()
print(service_line_by_hours)
# CYBERSECURITY has largest hours logged
```

```python
# 41.	Group by priority and count the tickets.

tickets_by_priority =  df.groupby('priority').size()
print(tickets_by_priority)

# groupby on priority column and then size() show the priority count of tickets
```

```python
# 42.	Group by client name and find the total billable amount.
client_name_by_total_billable_amount = df.groupby('client_name')['billable_amount_usd'].sum()
print(client_name_by_total_billable_amount)

# groupby using client_name then associate billable_amount_usd column and find the sum for each client.
# Pioneer Health Systems has the highest billable amount
```

```python
# 43.	Find the top 5 clients by number of tickets.
top_5_clints_by_tickets = df['client_name'].value_counts().head()
print(top_5_clints_by_tickets)

# Zephyr Semiconductors has highest tickets
```

```python
# 44.	Find the average CSAT score for each priority.
avg_csat_for_priority = df.groupby('priority')['csat_score'].mean()
print(avg_csat_for_priority)
#  csat score is almost same for each priority
```

```python
# 45.	Draw a bar chart of ticket count by service line.
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))
sns.countplot(data=df, y='serviceline', order = df['serviceline'].value_counts().index)
plt.title('Ticket Count by Service Line')
plt.xlabel('Ticket Count')
plt.ylabel('Service Line')
plt.show()
# here y axis is serviceline and y axis is the ticket count of that serviceline
# form the graph we can see the Application Developement has highest ticket count and application support has lowest
```
